## Container With Most WaterYou are given an integer array height of length n. There are n vertical lines drawn such that the two endpoints of the $i$-th line are $(i, 0)$ and $(i, \text{height}[i])$.  Find two lines that, together with the x-axis, form a container such that the container contains the most water.  Return the maximum amount of water a container can store.Note: You may not slant the container, and $n$ is at least 2.ExamplesExample 1:Input: height = [1, 8, 6, 2, 5, 4, 8, 3, 7]Output: 49Explanation: The above vertical lines are represented by array [1, 8, 6, 2, 5, 4, 8, 3, 7]. In this case, the max area of water the container can store is between the line at index 1 (height 8) and the line at index 8 (height 7).$$\text{Area} = \min(8, 7) \times (8 - 1) = 7 \times 7 = 49$$

```python
def max_water(heights):
    left=0
    right=len(heights)-1
    max_area=0
    while left<right:
        width=right-left
        current_area=width*min(heights[left],heights[right])
        max_area=max(current_area,max_area)
        if heights[left]<heights[right]:
            left+=1
        else:
            right-=1
    return max_area
heights=[1,7,9,5,0,1]
print(max_water(heights))
            
            
        
        
```
